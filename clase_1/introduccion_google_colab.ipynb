# Programación de Computadores en Python
## Introducción a Google Colab
#### Universidad Nacional de Colombia

Google Colab (Google Colaboratory) es una herramienta gratuita que permite escribir y ejecutar código Python directamente en el navegador.  
Está basada en **Jupyter Notebooks**. Es muy utilizada para **análisis de datos, machine learning y enseñanza**.

---



## 🌟 Generalidades
- Ejecuta código Python en la nube (no necesitas instalar nada).
- Guarda automáticamente en Google Drive.
- Puedes compartir tu trabajo como si fuera un documento de Google.
- Compatible con CPU, GPU y TPU.
- Basado en celdas: de **código** o **texto** (Markdown).



## 📝 Tipos de Celdas

### Celda de Código
Aquí puedes escribir y ejecutar código Python.

Ejemplo:
```python
print("Hola desde Google Colab!")
```

### Celda de Texto (Markdown)
Permite escribir texto con formato, títulos, listas, enlaces, imágenes, etc.

Ejemplo en Markdown:
```markdown
# Título 1
## Título 2
**Negrita** y *cursiva*.
```


````python

# Ejemplo de ejecución de código en Colab
nombre = "Colab"
print(f"Hola {nombre}!")
````

1. Haz clic dentro de la celda.

## ⚡ Uso de GPU y TPU

Para habilitar GPU o TPU:
1. Ve al menú `Entorno de ejecución (Runtime)` → `Cambiar tipo de entorno de ejecución`.
2. En **Acelerador de hardware**, selecciona **GPU** o **TPU**.
3. Guarda los cambios y vuelve a ejecutar las celdas.

Puedes comprobar si tienes GPU con:
```python
!nvidia-smi
```
